# 🎯 Orquestrador - Pipeline Completo Brasileirão 2026

Este notebook executa o pipeline completo na seguinte ordem:
1. **Bronze** → Extração e armazenamento de dados brutos
2. **Silver** → Limpeza e padronização dos dados
3. **Gold** → Agregação e métricas de negócio

> **Nota**: Se uma camada falhar, as seguintes não serão executadas.

```python
# Import mssparkutils (Fabric)
import mssparkutils
from datetime import datetime

# Configuração
notebooks = {
    "bronze": "bronze/bronze_classificacao",
    "silver": "silver/silver_classificacao", 
    "gold": "gold/gold_classificacao"
}

print(f"🕐 Início: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print("=" * 60)
```

```python
# Executar pipeline completo
resultados = {}

for camada, notebook_path in notebooks.items():
    try:
        print(f"\n▶ Executando: {camada.upper()}")
        print("-" * 40)
        
        # Executar notebook
        mssparkutils.notebook.run(notebook_path)
        
        resultados[camada] = "✅ Sucesso"
        print(f"✅ {camada.upper()} concluído")
        
    except Exception as e:
        resultados[camada] = f"❌ Erro: {str(e)}"
        print(f"❌ {camada.upper()} falhou: {e}")
        break  # Para se uma camada falhar

print("\n" + "=" * 60)
print("📊 RESUMO DO PIPELINE")
print("=" * 60)
for camada, status in resultados.items():
    print(f"  {camada.upper()}: {status}")
```
